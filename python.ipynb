```python
import pandas as pd

# The name of large file
file_path = 'Loan_status_2007-2020Q3.csv' 

# We read 100,000 rows at a time to stay efficient
chunk_size = 100000 
total_rows = 0
total_loan_amount = 0

print("Starting the audit of the LendingClub Ledger...")

for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
    total_rows += len(chunk)
    # Adding up the 'loan_amnt' column to see the total book value
    total_loan_amount += chunk['loan_amnt'].sum()
    print(f"Processed {total_rows:,} rows...")

print("--- FINAL AUDIT REPORT ---")
print(f"Total Loan Records: {total_rows:,}")
print(f"Total Portfolio Value: ${total_loan_amount:,.2f}")
```

```python
import pandas as pd

file_path = 'Loan_status_2007-2020Q3.csv'
chunk_size = 100000
summary_list = []

print("📊 Generating the Risk Summary (Numeric Only)...")

for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
    # We explicitly define the columns and math to avoid the error
    chunk_stats = chunk.groupby('grade').agg(
        Total_Loan_Amount=('loan_amnt', 'sum'),
        Avg_Interest_Rate=('int_rate', 'mean'),
        Loan_Count=('loan_amnt', 'count')
    ).reset_index()
    
    summary_list.append(chunk_stats)

# Final aggregation of all chunks
final_summary = pd.concat(summary_list).groupby('grade').agg({
    'Total_Loan_Amount': 'sum',
    'Avg_Interest_Rate': 'mean',
    'Loan_Count': 'sum'
}).reset_index()

# Save the final CSV
final_summary.to_csv('LendingClub_Risk_Summary.csv', index=False)

print("✅ Success! Check your folder for 'LendingClub_Risk_Summary.csv'")
print(final_summary)
```

```python
import pandas as pd
import numpy as np

file_path = 'Loan_status_2007-2020Q3.csv'
chunk_size = 100000
summary_list = []

print("📊 Generating the Risk Summary (with Data Cleaning)...")

for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
    # Step 1: Clean the data - Force these columns to be numbers
    chunk['loan_amnt'] = pd.to_numeric(chunk['loan_amnt'], errors='coerce')
    chunk['int_rate'] = pd.to_numeric(chunk['int_rate'], errors='coerce')
    
    # Step 2: Perform the math
    chunk_stats = chunk.groupby('grade').agg({
        'loan_amnt': 'sum',
        'int_rate': 'mean',
        'id': 'count'
    }).reset_index()
    
    summary_list.append(chunk_stats)

# Final aggregation
final_summary = pd.concat(summary_list).groupby('grade').agg({
    'loan_amnt': 'sum',
    'int_rate': 'mean',
    'id': 'sum'
}).reset_index()

# Final Polish
final_summary.columns = ['Loan_Grade', 'Total_Loan_Amount', 'Avg_Interest_Rate', 'Loan_Count']
final_summary.to_csv('LendingClub_Risk_Summary.csv', index=False)

print("✅ Success! The Summary is ready.")
print(final_summary)
```

```python
import pandas as pd

file_path = 'Loan_status_2007-2020Q3.csv'
chunk_size = 100000
summary_list = []

print("📊 Cleaning interest rates and finalizing report...")

for chunk in pd.read_csv(file_path, chunksize=chunk_size, low_memory=False):
    # Step 1: Strip the '%' and convert to a decimal/float
    if chunk['int_rate'].dtype == 'object':
        chunk['int_rate'] = chunk['int_rate'].str.rstrip('%').astype(float)
    
    # Step 2: to Ensure loan_amnt is a number
    chunk['loan_amnt'] = pd.to_numeric(chunk['loan_amnt'], errors='coerce')
    
    # Step 3: Aggregate
    chunk_stats = chunk.groupby('grade').agg(
        Total_Loan_Amount=('loan_amnt', 'sum'),
        Avg_Interest_Rate=('int_rate', 'mean'),
        Loan_Count=('loan_amnt', 'count')
    ).reset_index()
    
    summary_list.append(chunk_stats)

# Combine and save
final_summary = pd.concat(summary_list).groupby('grade').agg({
    'Total_Loan_Amount': 'sum',
    'Avg_Interest_Rate': 'mean',
    'Loan_Count': 'sum'
}).reset_index()

final_summary.to_csv('LendingClub_Risk_Summary_FINAL.csv', index=False)
print("✅ Success! 'LendingClub_Risk_Summary_FINAL.csv' is ready for Tableau.")
print(final_summary)
```
